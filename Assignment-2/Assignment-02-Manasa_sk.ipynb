Create an array $x$ with 100 points in the interval $[-1,1]$ (the points must be uniformly spaced). Then create another array $y$. 

Where $y = x^2 sin(x^2) + x$. 

Try to plot $y$ vs $x$ using matplotlib.

## Problem: 7 
Create a function which creates an $n×n$ array with $(i , j)$ -entry equal to $i+ j$

```python
# Write your code here

import numpy as np

def create_array(n):
    i = np.arange(n)
    j = np.arange(n)
    return i[:, None] + j

create_array(4)
```

```python
def create_sum_array(n):
 
    # Initialize an empty list to store the rows of the array
    array = []
    
    # Iterate through each row index 'i' from 0 to n-1
    for i in range(n):
        # Create a new list for the current row
        row = []
        # Iterate through each column index 'j' from 0 to n-1
        for j in range(n):
            # The value at this position is the sum of the row index (i) and column index (j)
            row.append(i + j)
        # Add the completed row to the main array
        array.append(row)
        
    # Return the final array
    return array

# Example Usage:
n_value = 4
result_array = create_sum_array(n_value)

# Print the array in a readable format
print(f"An {n_value}x{n_value} array with (i, j)-entry equal to i + j:")
for row in result_array:
    print(row)
```

# Problem: 8
Evaluate $cos$ and $sin$ on the interval $[0,1]$  with 1000 points and then stack the results into a tall array with rows being the $(cos(x),sin(x))$ entries.

```python
import numpy as np

# Generate 1000 points on the interval [0, 1]
x_values = np.linspace(0, 1, 1000)

# Evaluate cos(x) and sin(x)
cos_values = np.cos(x_values)
sin_values = np.sin(x_values)

# Stack the results into a tall array with rows as (cos(x), sin(x))
result_array = np.vstack((cos_values, sin_values)).T

print(result_array)

# The result_array now holds the (cos(x), sin(x)) pairs in a 1000x2 shape.
```

```python
import numpy as np

# 1. Create 1000 points between 0 and 1
x = np.linspace(0, 1, 1000)

# 2. Compute cos(x) and sin(x)
cos_x = np.cos(x)
sin_x = np.sin(x)

# 3. Stack as rows (cos(x), sin(x))
result = np.column_stack((cos_x, sin_x))

print(result)
# print(result.shape)
```
