### Problem-7

A batch of 500 microchips is produced. The probability that a single microchip is defective is 0.02.
Let X be the number of defective microchips in a random sample of 20 microchips from this batch. Assume the sample is taken with replacement, or the batch is large enough that we can model this as independent trials.

(a) What is the probability that exactly 3 microchips in the sample are defective?
(b) What is the probability that at least 2 microchips in the sample are defective?

###### no of microchipd- 500
###### microchip_singleerror - 0.02
###### x_defectivechip_per 20 
###### 2 mic defective 10 -0.2 

#### Binomial formula

$$
P(X=k) = \binom{n}{k} p^k (1-p)^{n-k}
$$

````python
import math

n = 20
p = 0.02

# (a) Exactly 3 defective
prob_3 = math.comb(n, 3) * (p**3) * ((1 - p)**(n - 3))

# (b) At least 2 defective
prob_0 = (1 - p)**n
prob_1 = math.comb(n, 1) * p * ((1 - p)**(n - 1))
prob_at_least_2 = 1 - prob_0 - prob_1

print("P(X = 3) =", prob_3)
print("P(X ≥ 2) =", prob_at_least_2)
````

### Problem-11

The lifespan of a certain electronic component, due to manufacturing variations, is normally distributed with a mean of 2500 hours and a standard deviation of 300 hours.

(a) A customer requires components that last at least 2000 hours. What percentage of manufactured components meet this requirement?

(b) The manufacturer wants to provide a warranty such that only 1% of components fail before the warranty period ends. What should be the warranty period (in hours)?

(c) If a batch of 10,000 components is produced, how many components are expected to have a lifespan between 2200 and 2800 hours?

\(z=\frac{x-\mu }{\sigma }=\frac{2000-2500}{300}=-1.6667\)

## Problem Statement

#### Electronic Component Lifespan Analysis

**Parameters:**
* Mean ($\mu$): 2500 hours
* Standard Deviation ($\sigma$): 300 hours

---

#### (a) Percentage of components lasting $\ge 2000$ hours

1. **Calculate the Z-score:**
   $$z = \frac{x - \mu}{\sigma} = \frac{2000 - 2500}{300} = -1.67$$

2. **Find the probability:**
   Using the standard normal table, $P(Z < -1.67) \approx 0.0475$.
   Since we need "at least" 2000 hours:
   $$P(Z \ge -1.67) = 1 - 0.0475 = 0.9525$$

**Result:** Approximately **95.25%** of components meet the requirement.

---

#### (b) Warranty period for 1% failure rate

We need to find the value $x$ where the area to the left is $0.01$.

1. **Find the Z-score for 1%:**
   From the Z-table, the area of $0.01$ corresponds to $z \approx -2.326$.

2. **Solve for $x$:**
   $$x = \mu + z\sigma$$
   $$x = 2500 + (-2.326 \times 300) = 1802.2$$

**Result:** The warranty period should be set to **1,802.2 hours**.

---

#### (c) Expected components between 2200 and 2800 hours

1. **Calculate Z-scores:**
   * For 2200: $z_1 = \frac{2200 - 2500}{300} = -1.0$
   * For 2800: $z_2 = \frac{2800 - 2500}{300} = 1.0$

2. **Calculate Probability:**
   According to the Empirical Rule (68-95-99.7), the area between $\pm 1\sigma$ is approximately **0.6827**.

3. **Calculate Expected Value:**
   $$E = 10,000 \times 0.6827 = 6,827$$

**Result:** Approximately **6,827 components** are expected to fall in this range.



````python
from scipy.stats import norm

mu = 2500
sigma = 300

prob_a = 1 - norm.cdf(2000, mu, sigma)
print(prob_a)

warranty_b = norm.ppf(0.01, mu, sigma)
print(warranty_b)

prob_c = norm.cdf(2800, mu, sigma) - norm.cdf(2200, mu, sigma)
expected_c = 10000 * prob_c

print(prob_c)
print(expected_c)
````

### Problem-14

A grayscale image can be represented as a matrix where the dimensions are (height, width). 
Let's consider a simplified scenario with two grayscale images, $A$ and $B$, both represented as 10×10 matrices where each element represents pixel intensity. Pixel intensity is an integer in the range [0-255].

You want to create a blended image $C$ by combining $A$ and $B$ such that $C=w_1 A+w_2 B$. This represents a weighted average blend.
Note that each element in the resultant matrix $C$ should also fall within the range [0, 255]. 

write the code of the function `blend_images(image_A, image_B, weight_A, weight_B)` that takes two matrices and two scalar weights and produces another matrix as an output by performing the above operation.

````python
### Write your code here
import numpy as np

def blend_images(image_A, image_B, weight_A, weight_B):
    # Convert inputs to numpy arrays
    A = np.array(image_A, dtype=float)
    B = np.array(image_B, dtype=float)
    
    # Weighted blending
    C = weight_A * A + weight_B * B
    
    # Clip values to stay within [0, 255]
    C = np.clip(C, 0, 255)
    
    # Convert back to integer pixel values
    return C.astype(np.uint8)
````

````python
A = np.random.randint(0, 256, (10, 10))
B = np.random.randint(0, 256, (10, 10))

C = blend_images(A, B, 0.6, 0.4)
print(C)
````

### Problem-15

Calculate the eigen values and eigen vectors of the following matrices.

$A = \begin{bmatrix}
5 & 2 \\
2 & 2
\end{bmatrix}$      

$B = \begin{bmatrix}
1 & 1 & 0\\
0 & 2 & 0\\
-1 & 1 & 3
\end{bmatrix}$

````python
import numpy as np

A = np.array([[5,2],[2,2]])
B = np.array([[1,1,0],[0,2,0],[-1,1,3]])

eig_A = np.linalg.eig(A)
eig_B = np.linalg.eig(B)

print("Eigenvalues of A:", eig_A[0])
print("Eigenvectors of A:\n", eig_A[1])

print("\nEigenvalues of B:", eig_B[0])
print("Eigenvectors of B:\n", eig_B[1])
````
