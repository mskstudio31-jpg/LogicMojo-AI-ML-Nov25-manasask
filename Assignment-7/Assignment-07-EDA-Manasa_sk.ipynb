```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Banking_churn_prediction.csv")

# show first rows
print(df.head())

```

```python
# dataset shape
print("Dataset Shape:", df.shape)

# column names
print("\nColumns:")
print(df.columns)

# data types
print("\nData Types:")
print(df.dtypes)

# summary statistics
print("\nSummary Statistics:")
print(df.describe())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print(df['churn'].value_counts())

sns.countplot(x='churn', data=df)
plt.title("Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['current_balance'], bins=30)
plt.title("Current Balance Distribution")
plt.show()
```

```python
sns.countplot(x='gender', hue='churn', data=df)
plt.title("Gender vs Churn")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='occupation', hue='churn', data=df)
plt.xticks(rotation=45)
plt.title("Occupation vs Churn")
plt.show()
```

```python
plt.figure(figsize=(10,8))
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=False, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.boxplot(x='churn', y='current_balance', data=df)
plt.title("Current Balance vs Churn")
plt.show()
```

```python
sns.countplot(x='dependents', data=df)
plt.title("Dependents Distribution")
plt.show()
```

```python
top_cities = df['city'].value_counts().head(10)

top_cities.plot(kind='bar')
plt.title("Top Cities")
plt.show()
```
