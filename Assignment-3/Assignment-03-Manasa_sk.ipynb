#### 11. Add a column that calculates the percent change between the 2010 Census and the 2017 Population Estimate

```python
# Write your code here

pop['Percent_Change_2010_2017'] = (
    (pop['2017PopEstimate'] - pop['2010Census'])
    / pop['2010Census']
) * 100

pop[['2010Census', '2017PopEstimate', 'Percent_Change_2010_2017']].head()
```

```python
# Write your code here
state_pop = pop.groupby('State')[['2010Census', '2017PopEstimate']].sum()

state_pop['Percent_Change_2010_2017'] = (
    (state_pop['2017PopEstimate'] - state_pop['2010Census'])
    / state_pop['2010Census']
) * 100

print(state_pop)
state_pop.sort_values('Percent_Change_2010_2017', ascending=False).head()


```
